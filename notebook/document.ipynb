```python
## Create multiple txt files

import os

# Create folder
os.makedirs("../data/text_files", exist_ok=True)

# Dictionary containing filename and content
sample_txt = {
    "rag.txt": """
Retrieval-Augmented Generation (RAG)
-----------------------------------
RAG combines retrieval systems with LLMs to generate accurate responses.
It retrieves relevant information from external sources before generating answers.
""",

    "ml.txt": """
Machine Learning
----------------
Machine Learning is a subset of Artificial Intelligence.
It enables systems to learn patterns from data and make predictions.
Common types include supervised, unsupervised, and reinforcement learning.
""",

    "nlp.txt": """
Natural Language Processing
---------------------------
NLP is a field of AI focused on understanding and processing human language.
Applications include chatbots, translation systems, and sentiment analysis.
"""
}

# Create and write multiple files
for filename, content in sample_txt.items():

    file_path = f"../data/text_files/{filename}"

    with open(file_path, "w", encoding="utf-8") as file:
        file.write(content)

print("All text files created successfully!")
```

```python
## TextLoader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/sample.txt", encoding="utf-8")
document=loader.load()
print(document)
```

```python
## Directory Loader

from langchain_community.document_loaders import DirectoryLoader
from langchain_community.document_loaders import TextLoader

# Load all txt files from directory
dir_loader = DirectoryLoader(
    "../data/text_files",
    glob="*.txt",                     # Match all .txt files
    loader_cls=TextLoader,           # Loader class
    loader_kwargs={"encoding": "utf-8"}  # Extra arguments for TextLoader
)

# Load documents
documents = dir_loader.load()

# Print loaded documents
print(documents)
```

```python
## pdf loader
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

# Load PDF using PyPDFLoader

dir_loader = DirectoryLoader(
    "../data/pdf",
    glob="*.pdf",                     # Match all .txt files
    loader_cls=PyMuPDFLoader,           # Loader class  
)

pdf_documents = dir_loader.load()
print(pdf_documents)
```
